```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd

def extract_players_vs_time(data):
    data = data.replace("M", "")
    data = data.replace("L", "")
    data = data.split(' ')
    data = list(filter(lambda a: a != "", data))# remove all instances of " " from the list
    data = np.array(data).astype(float)
    x = data[0:-2:2] # SyntaxError: invalid syntax
    y = data[1:-1:2]
    print(f'{len(x) = }')
    return x, y
    
def plot_players_vs_time(x, y, name):
    plt.figure(dpi=100)
    plt.title(f"Players vs Time for {name}")
    plt.plot(x, y)
    plt.xlabel('Time (Days)')
    plt.ylabel('Players')
    plt.show()
    
games = pd.DataFrame()
```
